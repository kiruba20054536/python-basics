1. Find how many odd and even numbers are there in between 2 numbers.

```python
a=int(input("Enter a number:"))
b=int(input("Enter another number:"))

count1=0
count2=0
for i in range(a,b+1):
    if i%2==0:
        #print(i) # to print the even numbers
        count1+=1
    elif i%2!=0:
        #print(i)  # to print the odd numbers
        count2+=1
print("Number of even:",count1)
print("Number of odd:",count2)
```

4. get input of 10 numbers and find the average for that numbers.

```python
c=int(input("How many numbers?"))
a = []
for i in range(1, c+1):
    b = int(input(f"Enter number {i}:"))
    a.append(b)

length=len(a)
total = 0
for i in a:
    total += i
print("Average of numbers:",total/length)
```

14. Reverse a string using a for loop (no slicing). Python’s [ : :-1] shortcut is famous.

```python
original = "Kiruba"
reverse = ""
for char in original:
    reverse = char + reverse
print(reverse)
# reverse = char + reverse
# K       = K    + ""
# iK      = i    + K
# riK     = r    + iK
# uriK    = u    + riK
# buriK   = b    + uriK
# aburiK  = a    + buriK
# 'aburiK' will return!!
```

17. Find largest and smallest digit in a number

```python
num = input("Enter a number: ")
largest_number = 0
smallest_number = 9
for i in num: # 6,4,5,1
    i = int(i)
    if i>=largest_number: 
        largest_number = i # 6
        
    if i<= smallest_number:
        smallest_number = i # 4,5,1
        
print(f"largest number is {largest_number}, smallest number is {smallest_number}.")
```
